## Task 6
What is the main difference between a tuple and a list?

**Answer:** A list is *mutable*: its items can be changed, added or removed after creation (e.g. `marks[0] = 80`, `marks.append(66)`). A tuple is *immutable*: once created it cannot be changed, and trying to do so raises a `TypeError`. Tuples are used for fixed data such as `(rows, cols)`.

```python
# Demonstration
my_list = [1, 2, 3]
my_list[0] = 10
print("List after change:", my_list)

my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 10
except TypeError as e:
    print("Tuple error:", e)
```

## Task 10
`caught_speeding(speed, is_birthday)`: ≤ 60 → "No Ticket", 61–80 → "Small Ticket", ≥ 81 → "Big Ticket". On a birthday, allow 5 more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5      # birthday: 5 extra allowed

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10:", caught_speeding(81, True))    # Small Ticket
print("Q10:", caught_speeding(81, False))   # Big Ticket
print("Q10:", caught_speeding(65, True))    # No Ticket
print("Q10:", caught_speeding(65, False))   # Small Ticket
```
