## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3: Randomness and reproducibility
np.random.seed(2026)                                  # (a)
print("T3a: one random number in [0, 1) =", np.random.rand())

normal_25 = np.random.randn(25)                       # (b) standard normal
print("T3b: 25 standard-normal numbers =\n", np.round(normal_25, 4))

np.random.seed(42)                                    # (c) 10,000 Ludo dice
rolls = np.random.randint(1, 7, 10_000)               # high is exclusive, so 7 gives faces 1 to 6
faces, counts = np.unique(rolls, return_counts=True)
print("T3c: faces =", faces)
print("T3c: counts =", counts)

p_six = counts[faces == 6][0] / rolls.size            # (d)
print("T3d: P(six) estimated =", round(p_six, 4), "| theoretical 1/6 =", round(1/6, 4),
      "| difference =", round(abs(p_six - 1/6), 4))

is_six = rolls == 6                                   # (e) two sixes in a row, no loop
double_sixes = np.sum(is_six[:-1] & is_six[1:])       # compare each roll with the next one
# (overlapping pairs: three sixes in a row would count as 2 pairs)
print("T3e: times two sixes came in a row =", double_sixes)
print("T3e: expected about 10,000 x (1/6)^2 =", round(10_000 / 36, 1))
```

**Interpretation (T6):** Every unit grew between 2017 and 2023, but at very different speeds. So AJK's population rose by only about 7 % in six years (4.05 M to 4.33 M), whereas the national picture is rapid growth of roughly 2.3 to 3.2 % per year. ## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: Climate of five cities
print("T7a: shape of tmax =", tmax.shape, "(5 cities x 12 months)")          # (a)

annual_mean = tmax.mean(axis=1)                                            # (b) axis=1 -> one value per city
print("T7b: annual mean max temperature (deg C):")
print(np.column_stack((cities, np.round(annual_mean, 2))))

hottest = months[tmax.argmax(axis=1)]                                      # (c)
print("T7c: hottest month of each city:")
print(np.column_stack((cities, hottest, tmax.max(axis=1))))

print("T7d: 5-city average for each month (deg C):")                      # (d) axis=0 -> one value per month
print(np.column_stack((months, np.round(tmax.mean(axis=0), 2))))

above35 = (tmax > 35).sum(axis=1)                                          # (e) mask, then count per city
print("T7e: number of months above 35 deg C:")
print(np.column_stack((cities, above35)))

lahore_f = tmax[0] * 9/5 + 32                                              # (f)
print("T7f: Lahore monthly max in deg F =", np.round(lahore_f, 1))

monsoon = rain[:, 6:9].sum(axis=1)                                         # (g) Jul, Aug, Sep = columns 6, 7, 8
print("T7g: monsoon (Jul-Sep) rain (mm):")
print(np.column_stack((cities, np.round(monsoon, 1))))

print("T7h: wettest monsoon city =", cities[monsoon.argmax()],
      "with", round(monsoon.max(), 1), "mm")                               # (h)
monsoon_pct = monsoon / rain.sum(axis=1) * 100
print("T7h: monsoon share of yearly rain (%):")
print(np.column_stack((cities, np.round(monsoon_pct, 1))))

# helper numbers used in the interpretation below
print("T7 extra: yearly rain (mm):", np.column_stack((cities, np.round(rain.sum(axis=1), 1))).tolist())
```

**Interpretation (T7):** What surprised me most is how different the monsoon is across cities. Islamabad gets 763.0 mm in Jul-Sep, about 43 times more than Quetta's 17.9 mm, and Quetta receives only 6.0 % of its yearly rain in the monsoon (its rain comes in winter, Jan-Mar), while Karachi gets 73.8 % of its small total (318.0 mm a year) in those three months. Also, Karachi has the highest annual mean maximum (32.51 deg C) even though its hottest month (35.7 deg C) is cooler than Peshawar's June (39.6 deg C), because Karachi stays warm all winter.

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8: Inflation FY2024 vs FY2025
avg = cpi.mean(axis=1)                                                     # (a) axis=1 -> one value per year
print("T8a: average inflation FY2024 =", round(avg[0], 2), "% | FY2025 =", round(avg[1], 2), "%")

print("T8b: FY2024 months above 25 % =", cpi_months[cpi[0] > 25])          # (b) mask

drop = cpi[0] - cpi[1]                                                     # (c) FY2024 - FY2025
print("T8c: drop in inflation per month (percentage points):")
print(np.column_stack((cpi_months, np.round(drop, 1))))

print("T8d: biggest drop in", cpi_months[drop.argmax()],
      "=", round(drop.max(), 1), "percentage points")                      # (d)

label = np.where(cpi[1] < 5, "Low",                                        # (e) nested np.where
                 np.where(cpi[1] < 10, "Moderate", "High"))
print("T8e: FY2025 labels:")
print(np.column_stack((cpi_months, cpi[1], label)))
```

**Interpretation (T8):** Inflation fell sharply from FY2024 to FY2025: the average dropped from 26.22 % to 4.87 %, with a fall of between 17.0 and 25.9 percentage points in every month (biggest in January: 28.3 % to 2.4 %). Prices were still rising in FY2025, because every value is above zero, but much more slowly: by April the year-on-year rate was only 0.3 %. This is disinflation (prices rising more slowly), not deflation (prices falling).

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: Grace marks
after = np.clip(marks + 5, 0, 100)                                         # (a) +5, capped at 100
print("T9a: marks before =", marks)
print("T9a: marks after  =", after)

pass_pct = (after >= 50).mean() * 100                                      # (b)
print("T9b: pass percentage after grace =", pass_pct, "%")
print("T9b: (pass percentage before grace =", (marks >= 50).mean() * 100, "%)")

only_grace = np.where((marks < 50) & (after >= 50))[0]                     # (c) failed before, pass after
print("T9c: indices of students who passed ONLY because of grace =", only_grace)
print("T9c: their marks before -> after =", marks[only_grace], "->", after[only_grace])

z = (after - after.mean()) / after.std()                                   # (d) z-score (population std, np default)
print("T9d: mean after grace =", after.mean(), "| std =", round(after.std(), 4))
print("T9d: z-scores =", np.round(z, 3))
print("T9d: check, mean of z =", round(z.mean(), 10), "| std of z =", round(z.std(), 10))
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# T10: An image is an array
img = np.zeros((8, 8), dtype=int)                                          # (a) draw the "7"
img[1, 1:7] = 255          # top bar
img[2:7, 5] = 255          # diagonal-ish stem going down
img[4, 3:7] = 128          # grey middle bar (overwrites pixel [4, 5])
print("T10a: the 7 =\n", img)

flipped = img[:, ::-1]                                                     # (b) left-right
print("T10b: flipped left-right =\n", flipped)

inverted = 255 - img                                                       # (c)
print("T10c: inverted =\n", inverted)

crop = img[1:5, 3:7]                                                       # (d) rows 1-4, columns 3-6
print("T10d: crop rows 1-4, cols 3-6 =\n", crop)

bright = np.clip(img + 80, 0, 255)                                         # (e)
print("T10e: brightened by 80 (capped at 255) =\n", bright)

print("T10f: shape =", img.shape, "| dtype =", img.dtype,
      "| white pixels (=255) =", (img == 255).sum())                       # (f)

# Optional: see it as a picture (sneak peek of Lab 4)
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 4, figsize=(10, 3))
ax[0].imshow(img, cmap="gray", vmin=0, vmax=255);      ax[0].set_title("Original 7")
ax[1].imshow(flipped, cmap="gray", vmin=0, vmax=255);  ax[1].set_title("Flipped")
ax[2].imshow(inverted, cmap="gray", vmin=0, vmax=255); ax[2].set_title("Inverted")
ax[3].imshow(bright, cmap="gray", vmin=0, vmax=255);   ax[3].set_title("Brighter +80")
plt.show()
```
